# Regressão Linear

### Exemplos de Dados

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
x = 2 * np.random.rand(100, 1)
y = 4 + 3 * x + np.random.randn(100, 1)
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
plt.scatter(x, y)
```

```python
x_b = np.c_[np.ones((100,1)),x] # adiciona x0 = 1 a cada instância
```

```python
theta = np.linalg.inv(x_b.T.dot(x_b)).dot(x_b.T).dot(y)
theta
```

```python
y_chapeu = theta[0] + x * theta[1]

fig, ax = plt.subplots(figsize=(8, 6))
plt.plot(x, y_chapeu, "r-")
plt.scatter(x, y)
```

### Gradient Descent

```python
eta = 0.1 # taxa de aprendizado
n_interactions = 1000 # número de interações
m = 100 # número de amostras
```

```python
theta = np.random.randn(2,1) # inicializa theta aleatoriamente
theta
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
plt.scatter(x, y)

for interaction in range(n_interactions):
		gradients = 2/m * x_b.T.dot(x_b.dot(theta) - y) # calcula o gradiente
		theta = theta - eta * gradients                 # atualiza theta
		y_chapeu = theta[0] + x * theta[1]              # plota a linha de regressão com transparência crescente
  
		ax.plot(x, y_chapeu, alpha=0.001+(interaction / n_interactions), color="r") # plota a linha de regressão com transparência crescente
```

### Regressão Linear com Gradiente Descendente

```python
# Aplicando o Gradiente Descendente para Regressão Linear

data = 2 * np.random.rand(100, 1) # gerando dados aleatórios
target = 4 + 3 * data + np.random.randn(100, 1) # gerando o alvo com ruído

eta = 0.1 # taxa de aprendizado
n_interactions = 1000 # número de interações
m = 100 # número de amostras

theta = np.random.randn(2,1) # inicializa theta aleatoriamente

# Aplicando o Gradiente Descendente na Regressão Linear

fig, ax = plt.subplots(figsize=(8, 6))
plt.scatter(data, target)

for interaction in range(n_interactions):
		gradients = 2/m * x_b.T.dot(x_b.dot(theta) - target) # calcula o gradiente
		theta = theta - eta * gradients                      # atualiza theta
		y_chapeu = theta[0] + data * theta[1]               # plota a linha de regressão com transparência crescente
	
		ax.plot(data, y_chapeu, alpha=0.001+(interaction / n_interactions), color="r") # plota a linha de regressão com transparência crescente

```
