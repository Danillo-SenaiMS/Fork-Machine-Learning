### Regressão Polinomial

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib
```

```python
m = 100 # número de amostras
x = 6 * np.random.rand(m,1)-3 # gera 100 amostras aleatórias entre -3 e 3
y = 0.5 * x**2 + x + 2 * np.random.randn(m,1) # gera os valores de y com ruído
```

```python
plt.scatter(x, y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
# Cria um objeto PolynomialFeatures com grau 2 e sem bias
poly_features = PolynomialFeatures(degree=2, include_bias=False)
x_poly = poly_features.fit_transform(x)
```

```python
from sklearn.linear_model import LinearRegression

# Cria um modelo de regressão linear
lin_reg = LinearRegression()
lin_reg.fit(x_poly, y) # Treina o modelo com os dados polinomiais

# Faz previsões com o modelo treinado
x_new = np.linspace(-3, 3, 100).reshape(100,1)
x_new_poly = poly_features.transform(x_new) # Transforma os novos dados em polinomiais
y_new = lin_reg.predict(x_new_poly) # Faz previsões com os novos dados
```

```python
# Intercepto do modelo de regressão polinomial
lin_reg.intercept_
```

```python
# Coefficients of the polynomial regression model
lin_reg.coef_
```

```python
plt.scatter(x, y)
plt.scatter(x_new, y_new, color='red', linewidth=2, label='Regressão Polinomial')
```
